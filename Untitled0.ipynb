```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

study_hours = [2, 3, 4, 5, 6, 7, 8, 9, 10,
               3, 5, 6, 8, 9, 10]

marks = [45, 50, 55, 60, 62, 68, 72, 75, 80,
         48, 58, 65, 70, 78, 85]

data = pd.DataFrame({
    "Students": range(0, 15),
    "Study Hours (X)": study_hours,
    "Marks (Y)": marks
})
```

```python
print(data)
```

```python
study_hour_mean = np.mean(data["Study Hours (X)"])
study_mark_mean = np.mean(data["Study Hours (X)"])
print("Number of students:", len(data))
print("Average study hours Mean:", round(study_hour_mean, 3))
print("Average marks:", round(study_mark_mean, 3))
```

```python

pearson_correlation = data["Study Hours (X)"].corr(
    data["Marks (Y)"], method="pearson"
)

print("Pearson Correlation Coefficient:", round(pearson_correlation, 4))
```

```python
spearman_correlation = data["Study Hours (X)"].corr(
    data["Marks (Y)"], method="spearman"
)

print("Spearman Rank Correlation Coefficient:", round(spearman_correlation, 4))
```

```python
import numpy as np
from scipy import stats

marks = [68, 72, 75, 69, 71, 74, 66, 73, 70, 76]

sample_mean = np.mean(marks)

print("Marks::", marks)
print("Number of students:", len(marks))
print("Sample mean:", sample_mean)
```

```python
#Testing Hypotheses here

claimed_mean = 70

t_statistic, p_value = stats.ttest_1samp(
    marks, popmean=claimed_mean
)

print("Sample Mean:", sum(marks) / len(marks))
print("t-statistic:", round(t_statistic, 4))
print("p-value:", round(p_value, 4))

#At 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
# Questions
# 1.	Why is a t-test used when the population standard deviation is unknown?
# Ans: A t-test is used when the population standard deviation is unknown because the sample standard deviation is used to estimate population variability. It accounts for the uncertainty in this estimate and is suitable for testing a population mean using a small sample.

# 2.	What does the p-value indicate?
# Ans: The p-value indicates how compatible the observed data are with the null hypothesis. A p-value below 0.05 provides statistically significant evidence against the null hypothesis at the 5% significance level. In this practical, the p-value was 0.2004, so we fail to reject the null hypothesis.
```

```python
# Practical 2nd


import numpy as np
from scipy import stats

method_a = [65, 70, 68, 72, 66, 69, 71, 67]

method_b = [72, 75, 78, 74, 80, 77, 76, 79]

mean_a = np.mean(method_a)
mean_b = np.mean(method_b)

print("Method A marks:::", method_a)
print("Method B marks:::", method_b)
print("Number of students in Method A:::", len(method_a))
print("Number of students in Method B:::", len(method_b))
print("Mean marks for Method A:::", mean_a)
print("Mean marks for Method B:::", mean_b)
```

```python
t_statistic, p_value = stats.ttest_ind(
    method_a,
    method_b,
    equal_var=False
)

print("Mean of Method A:", np.mean(method_a))
print("Mean of Method B:", np.mean(method_b))
print("t-statistic:", round(t_statistic, 4))
print("p-value:", p_value)

# at the 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
# Questions for Practice
# 3. How does an independent t-test differ from a paired t-test?
# Ans: An independent-samples t-test is used to compare the means of two independent groups, whereas a paired t-test is used to compare two related sets of observations, such as measurements taken from the same individuals before and after a treatment.
# 4. Why use equal_var=False?
# Ans: We use equal_var=False to perform Welch's independent-samples t-test. This test does not assume equal population variances and is suitable when the variances or sample sizes of the two groups may differ.
```

```python
# Practical 3: Paired-Samples t-Test


import numpy as np
from scipy import stats

before = [60, 62, 58, 65, 61, 63, 59, 64]

after = [65, 64, 62, 68, 66, 67, 63, 69]

mean_before = np.mean(before)
mean_after = np.mean(after)

print("Before training:::", before)
print("After training:::", after)
print("Number of students:::", len(before))
print("Mean before training:::", mean_before)
print("Mean after training:::", mean_after)
```

```python

#Testing Hypotheses here

t_statistic, p_value = stats.ttest_rel(after, before)
differences = np.array(after) - np.array(before)

print("Mean difference:::", np.mean(differences))
print("t-statistic:::", t_statistic)
print("p-value:::", p_value)

# At 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
#Practical 4: Chi-Square Goodness-of-Fit Test

import numpy as np
from scipy import stats

observed = [30, 20, 28, 22]
expected = [25, 25, 25, 25]

print("Observed frequencies:::", observed)
print("Expected frequencies:::", expected)
print("Total observed:::", sum(observed))
print("Total expected:::", sum(expected))
```

```python
# Testing the Hypothese here

observed = [30, 20, 28, 22]
expected = [25, 25, 25, 25]

chi_square, p_value = stats.chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-square statistic:::", chi_square)
print("P-value:::", p_value)
print("Degrees of freedom:::", len(observed) - 1)

# At 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Decision:: Reject the null hypothesis.")
else:
    print("Decision:: Fail to reject the null hypothesis.")
```

```python
# Q7. What is an expected frequency?
# An expected frequency is the number of observations expected in a category under the null hypothesis. In this practical, the expected frequency for each software package is 25 because 100 students are divided equally among four software packages.
# Q8. How does the Chi-Square Goodness-of-Fit Test differ from the Chi-Square Test of Independence?
# The Chi-Square Goodness-of-Fit Test compares observed frequencies with expected frequencies for one categorical variable. The Chi-Square Test of Independence examines whether two categorical variables are associated with each other.
```

```python
#PRACTICAL 5: Chi-Square Test of Independence

import numpy as np
from scipy import stats

table = np.array([
    [30, 20],
    [20, 30]
])

print("Observed frequency table:::")
print(table)

print("Grand total:::", table.sum())
print("Row totals:::", table.sum(axis=1))
print("Column totals:::", table.sum(axis=0))
```

```python

import numpy as np
from scipy import stats

table = np.array([
    [30, 20],
    [20, 30]
])

chi_square, p_value, df, expected = stats.chi2_contingency(
    table, correction=False
)

print("Expected frequencies:")
print(expected)

print("Chi-square statistic:", chi_square)
print("Degrees of freedom:", df)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the null hypothesis.")
else:
    print("Decision: Fail to reject the null hypothesis.")
```

```python
#Practical 7 — Correlation Analysis

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [45, 50, 52, 60, 65, 68, 75, 80]

print("Study hours:", x)
print("Marks:", y)
print("Number of students:", len(x))
```

```python


result = stats.linregress(x, y)

slope = result.slope
intercept = result.intercept
r_squared = result.rvalue ** 2
p_value = result.pvalue

study_hours = 5.5
predicted_marks = intercept + slope * study_hours

print("Slope:", slope)
print("Intercept:", intercept)
print("R-squared:", r_squared)
print("P-value:", p_value)
print("Predicted marks for 5.5 hours:", predicted_marks)

if p_value < 0.05:
    print("Decision: Reject the null hypothesis.")
else:
    print("Decision: Fail to reject the null hypothesis.")
```
